```python
import numpy as np

filename = input("Enter filename: ")

# load two columns
x, y = np.loadtxt(filename, unpack=True)

# statistics
print("x min:", x.min())
print("x max:", x.max())
print("x mean:", x.mean())

print("y min:", y.min())
print("y max:", y.max())
print("y mean:", y.mean())

# median
y_med = np.median(y)

# new array
y2 = (y - y_med)**2

print("y2 =", y2)
```

#### 5.9

```python
import imageio
import numpy as np
import matplotlib.pyplot as plt

# read image
img = imageio.imread("m101BW.jpg")

# display with color map
plt.imshow(img, cmap='jet')
plt.colorbar()
plt.show()

# contour plot
levels = np.linspace(np.min(img), np.max(img), 3)

plt.contour(img, levels=levels)
plt.show()

# keep highest 10% intensity
threshold = np.percentile(img, 90)

filtered = np.where(img >= threshold, img, threshold)

plt.imshow(filtered, cmap='jet')
plt.colorbar()
plt.show()
```

#### 5.10

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.loadtxt("ngc6946.txt")

# reshape cube
cube = data.reshape((40,50,70))

# middle position
x = 25
y = 35

z_profile = cube[:,x,y]

# statistics
print("min:", z_profile.min())
print("max:", z_profile.max())
print("mean:", z_profile.mean())
print("sum:", z_profile.sum())
print("std:", z_profile.std())

# plot
plt.plot(z_profile)

plt.text(0, z_profile.max()*0.9,
         f"mean={z_profile.mean():.2f}\nstd={z_profile.std():.2f}")

plt.xlabel("z")
plt.ylabel("Intensity")
plt.title("Z-profile")
plt.show()
```

#### 5.11

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.loadtxt("ngc6946.txt")
cube = data.reshape((40,50,70))

plane = cube[0,:,:]

mean = plane.mean()
std = plane.std()

plt.imshow(plane, origin="lower")
plt.colorbar()

levels = [mean, mean+std, mean+2*std]
plt.contour(plane, levels=levels, colors='red')

plt.show()
```

#### 5.12

```python
from scipy.constants import c

class Galaxy:
    """Class for redshift calculation of galaxies"""

    caH_0 = 3968.5
    c = c   # class variable

    def __init__(self, name, calciumHline=caH_0):
        self.name = name
        self.caH = calciumHline

    def redshift(self):
        z = (self.caH - self.caH_0)/self.caH
        return z

    def velocity(self):
        z = self.redshift()
        v = z * self.c / 1000   # convert m/s → km/s
        return v

# example
g = Galaxy("NGC1234", 4000)

print("Velocity:", g.velocity(), "km/s")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Ellipse

Nx = 400
Ny = 300

data = np.random.random((Ny,Nx))

fig, ax = plt.subplots()

ax.imshow(data, interpolation='none', origin="lower", cmap='gray')

plt.show()
```

```python
import numpy as np
from matplotlib.pyplot import figure, show
Nx = 400
Ny = 300
data = np.random.random((Ny,Nx))

figure, ax = plt.subplots()

ax.imshow(data, interpolation='none', origin="lower", cmap='gray')


# ellipse parametres
ellipse = Ellipse(
    (Nx/2, Ny/2),
    width=200,      # 2a
    height=100,     # 2b
    angle=30,
    edgecolor='red',
    facecolor='none',
    linewidth=2
)


ax.add_patch(ellipse)

plt.show()
```
