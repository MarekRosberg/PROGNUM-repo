```python
from math import *
from scipy.constants import G

masses = [1.9891e+30, 1.8986e+27, 5.6846e+26, 
          1.0243e+26, 8.6810e+25, 5.9736e+24, 
          4.8685e+24, 6.4185e+23, 3.3022e+23, 
          7.349e+22, 1.25e22]
radii  = [696340, 71492, 60268, 24764, 
          25559, 6378.1, 6051.8, 3396.2, 
          2439.7, 1738.1, 1195]
names  = ['Sun', 'Jupiter', 'Saturn', 'Neptune', 
          'Uranus', 'Earth', 'Venus', 'Mars', 
          'Mercury', 'Moon', 'Pluto'] #List of all values and names
MAIN = []
numbers = range(len(names)) #Range of names
numbers1 = range(len(masses)) #Range of masses
numbers2 = range(len(radii)) #Range of radii
for n, R, M in zip(numbers, numbers2, numbers1): #All restrictions and loops
            vesc = sqrt((2*G*masses[M])/radii[R]) #Calculating escape velocity
            print(f" {names[n]} = {vesc:6.2e}, m/s with mass, {masses[M]:6.2e} kg") #Printing with f string
            MAIN.append(vesc) #Adding values to MAIN
MAIN_dictionary = dict(zip(names, MAIN)) #Creating the dictionary
print(MAIN_dictionary)
```

```python
from scipy.constants import pi
import numpy as np
r = float(input("Radius of the cylinder:"))
h = float(input("Height of the cylinder:")) #Inputs for r and h
def cylinder(r,h):
    A = 2*pi*r**2 + 2*pi*r*h
    V = pi*r**2*h
    return A,V #Function of the cylinder
print("Area of the cylinder is:", cylinder(r,h)) 
a = np.array(cylinder(r,h)) #Array from calculated values
ratio = a[0]/a[1] #Calculating ratio from array
print("Ratio between area and volume is:", ratio)
```

```python
from math import *
import numpy as np
N = 5
x = []
for i in range(N):
    a = float(input(f"Value {i+1}:"))
    x.append(a) #Input of values into the list
y = np.array(x)
def mu(N, y): #Calculating the mean
    mu = np.sum(y)/N
    return mu
def stdev(N,y): #Calculating the Standard deviation
    m = mu(N, y)
    stdev = sqrt(np.sum((y-m)**2)/N)
    return stdev
std_oneline = sqrt(np.sum(y-(np.sum(y)/N)**2/N))
std = stdev(N, y)
print(f"Mean is {mu(N, y)}")
print(f"Standard Deviation is {stdev(N,y)}")
a = np.std(y) #Test of the standard deviation
print(f"Test of the standard deviation: {a}")
print(std_oneline)
```

```python
import numpy as np
distances = [18.49, 35.48, 35.48, 11.34, 7.12, 
             9.82, 23.66, 44, 66.7, 52.4, 55, 
             25.6, 26.19]
velocities = [2094.75, 2320.87, 1224.9, 872.13, 432.2, 751,
             1410, 2700, 5292, 2550, 5253, 2419, 2158]
d_array = np.array(distances)
v_array = np.array(velocities) #Creating arrays from lists

v_avg = np.average(velocities)
d_avg = np.average(distances) #Calculating the averages


b_cal = np.sum((d_array-d_avg)*(v_array-v_avg))/sum((d_array-d_avg)**2)
a_cal = v_avg - b_cal*d_avg #Calculating a and b

print("Slope of H0 is:", b_cal)
print("Intercept =", a_cal)
```
