```python
from matplotlib import pyplot as plt
import numpy as np
x = list(range(-5, 6)) #Range of the x-axis
y = np.cosh(x) #Function of the hyperbolic cosine

plt.plot(x, y, label="y = cosh(x)", color='black', marker='x')
plt.title("Hyperbolic Cosine", fontsize='15')
plt.xlabel("x", fontsize=12)
plt.ylabel("cosh(x)", fontsize=12)
plt.legend(fontsize=14) #All parameters for the graph
plt.xticks(range(-5, 6), fontsize=8)
plt.yticks(np.arange(-5, 80, 5), fontsize=8)
plt.show() #Display of the graph

x_2 = np.arange(-5, 6, 1)
y_2 = np.cosh(x_2) #Set up for the values of cosh

x_2a = np.array(x_2)
y_2a = np.array(y_2) #Array of all the values of cosh from -5 to 6

for i in range(len(x_2a)):
    print(f"For value {x_2a[i]}, cosh equals {y_2a[i]:.2f}") #All values printed
    
```
