```python
from math import *

#Angle input
x = int(input("x = "))
y = int(input("y = "))

#Evaluate angle
angle = math.atan2(y,x)

print("Angle =", angle)

#Write out each Quadrants
if 0 < angle < pi/2:
    print("Vector is in the 1st Quadrant")
    
elif pi/2 < angle < pi:
    print("Vector is in the 2nd Quarter")

elif -pi < angle < -pi/2:
    print("Vector is in the 3rd Quarter")
    
else: 
    print("Vector is in the 4th Quarter")
```

```python
night1 = [1,3,2,4,0,3,7,5,2,9,2,4,3]
night2 = [0,0,2,4,2,4,3,2,1,1,1,2,0]

set_night1 = set(night1)
set_night2 = set(night2)

intersect = list(set.intersection(set_night1, set_night2)) #Intersection of both sets
print("Values that intersect in both sets are", set(intersect))

if 5 in intersect:
    print("5 is present in the intersection")
else:
    print("5 is not present in the intersection")

if 5 in set(night1):
    print("5 is present in night1 sheet")
else:
    print("5 is not present in night1 sheet")
```
